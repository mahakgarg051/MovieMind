```python
def recommend_movies(user_id, n=5):

    # Find target user's cluster
    target_cluster = user_clusters[
        user_clusters['User_ID'] == user_id
    ]['Cluster'].iloc[0]

    # Get users from the same cluster
    cluster_users = user_clusters[
        user_clusters['Cluster'] == target_cluster
    ]['User_ID'].tolist()

    # Get cosine similarity with users in the same cluster
    cluster_similarity = user_similarity_df.loc[
        user_id,
        cluster_users
    ]

    # Remove target user
    cluster_similarity = cluster_similarity.drop(
        user_id,
        errors='ignore'
    )

    # Select top similar users
    similar_users = cluster_similarity.sort_values(
        ascending=False
    ).head(10)

    recommendation_data = []

    # Collect movies rated by similar users
    for similar_user, similarity_score in similar_users.items():

        user_ratings = df[
            df['User_ID'] == similar_user
        ]

        for _, row in user_ratings.iterrows():

            recommendation_data.append({
                'Movie_ID': row['Movie_ID'],
                'Movie_Title': row['Movie_Title'],
                'Rating': row['Rating'],
                'Similarity': similarity_score
            })

    recommendation_df = pd.DataFrame(recommendation_data)

    # Calculate weighted recommendation score
    recommendation_df['Weighted_Score'] = (
        recommendation_df['Rating'] *
        recommendation_df['Similarity']
    )

    # Remove movies already rated by target user
    already_rated = set(
        df[df['User_ID'] == user_id]['Movie_ID']
    )

    recommendation_df = recommendation_df[
        ~recommendation_df['Movie_ID'].isin(already_rated)
    ]

    # Group and sort recommendations
    final_recommendations = (
        recommendation_df
        .groupby(
            ['Movie_ID', 'Movie_Title'],
            as_index=False
        )
        .agg({
            'Weighted_Score': 'mean'
        })
        .sort_values(
            'Weighted_Score',
            ascending=False
        )
    )

    return final_recommendations.head(n)
```

```python
recommend_movies(
    user_id=157,
    n=5
)
```

```python
from sklearn.model_selection import train_test_split
```

```python
train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42
)

print("Training data:", train_df.shape)
print("Testing data:", test_df.shape)
```

```python
relevant_movies = test_df[
    test_df['Rating'] >= 7
]

relevant_movies.head()
```

```python
def precision_at_k(recommended_movies, relevant_movies, k=5):

    recommended = set(recommended_movies[:k])
    relevant = set(relevant_movies)

    if len(recommended) == 0:
        return 0

    return len(recommended & relevant) / len(recommended)
```

```python
def recall_at_k(recommended_movies, relevant_movies, k=5):

    recommended = set(recommended_movies[:k])
    relevant = set(relevant_movies)

    if len(relevant) == 0:
        return 0

    return len(recommended & relevant) / len(relevant)
```

```python
train_matrix = train_df.pivot_table(
    index='User_ID',
    columns='Movie_ID',
    values='Rating'
).fillna(0)

train_matrix.shape
```

```python
K = 4

train_kmeans = KMeans(
    n_clusters=K,
    random_state=42,
    n_init=10
)

train_clusters = train_kmeans.fit_predict(train_matrix)

train_user_clusters = pd.DataFrame({
    'User_ID': train_matrix.index,
    'Cluster': train_clusters
})

train_user_clusters.head()
```

```python
train_similarity = cosine_similarity(train_matrix)

train_similarity_df = pd.DataFrame(
    train_similarity,
    index=train_matrix.index,
    columns=train_matrix.index
)
```

```python
def recommend_from_train(user_id, n=5):

    if user_id not in train_matrix.index:
        return pd.DataFrame(
            columns=['Movie_ID', 'Movie_Title', 'Weighted_Score']
        )

    # Target user's cluster
    target_cluster = train_user_clusters[
        train_user_clusters['User_ID'] == user_id
    ]['Cluster'].iloc[0]

    # Same cluster users
    cluster_users = train_user_clusters[
        train_user_clusters['Cluster'] == target_cluster
    ]['User_ID'].tolist()

    # Similarity with same-cluster users
    similarities = train_similarity_df.loc[
        user_id,
        cluster_users
    ].drop(user_id, errors='ignore')

    # Top similar users
    similar_users = similarities.sort_values(
        ascending=False
    ).head(10)

    recommendation_data = []

    for similar_user, similarity_score in similar_users.items():

        ratings = train_df[
            train_df['User_ID'] == similar_user
        ]

        for _, row in ratings.iterrows():

            recommendation_data.append({
                'Movie_ID': row['Movie_ID'],
                'Movie_Title': row['Movie_Title'],
                'Rating': row['Rating'],
                'Similarity': similarity_score
            })

    recommendation_df = pd.DataFrame(recommendation_data)

    if recommendation_df.empty:
        return recommendation_df

    # Weighted recommendation score
    recommendation_df['Weighted_Score'] = (
        recommendation_df['Rating'] *
        recommendation_df['Similarity']
    )

    # Remove movies already rated in training data
    already_rated = set(
        train_df[
            train_df['User_ID'] == user_id
        ]['Movie_ID']
    )

    recommendation_df = recommendation_df[
        ~recommendation_df['Movie_ID'].isin(already_rated)
    ]

    # Final recommendations
    final_recommendations = (
        recommendation_df
        .groupby(
            ['Movie_ID', 'Movie_Title'],
            as_index=False
        )['Weighted_Score']
        .mean()
        .sort_values(
            'Weighted_Score',
            ascending=False
        )
    )

    return final_recommendations.head(n)
```

```python
test_recommendations = recommend_from_train(
    user_id=157,
    n=5
)

test_recommendations
```

```python
recommended_movies = test_recommendations['Movie_ID'].tolist()

relevant_movies = test_df[
    (test_df['User_ID'] == 157) &
    (test_df['Rating'] >= 7)
]['Movie_ID'].tolist()

print("Recommended Movies:", recommended_movies)
print("Relevant Movies:", relevant_movies)
```

```python
precision = precision_at_k(
    recommended_movies,
    relevant_movies,
    k=5
)

recall = recall_at_k(
    recommended_movies,
    relevant_movies,
    k=5
)

print("Precision@5:", precision)
print("Recall@5:", recall)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np
```

```python
def predict_rating(user_id, movie_id, n_users=10):

    if user_id not in train_matrix.index:
        return np.nan

    if movie_id not in train_matrix.columns:
        return np.nan

    target_cluster = train_user_clusters[
        train_user_clusters['User_ID'] == user_id
    ]['Cluster'].iloc[0]

    cluster_users = train_user_clusters[
        train_user_clusters['Cluster'] == target_cluster
    ]['User_ID'].tolist()

    similarities = train_similarity_df.loc[
        user_id,
        cluster_users
    ].drop(user_id, errors='ignore')

    # Users who actually rated this movie
    movie_ratings = train_df[
        (train_df['Movie_ID'] == movie_id) &
        (train_df['User_ID'].isin(similarities.index))
    ]

    if movie_ratings.empty:
        return np.nan

    movie_ratings = movie_ratings.copy()

    movie_ratings['Similarity'] = movie_ratings['User_ID'].map(
        similarities
    )

    movie_ratings = movie_ratings[
        movie_ratings['Similarity'] > 0
    ]

    if movie_ratings.empty:
        return np.nan

    movie_ratings = movie_ratings.sort_values(
        'Similarity',
        ascending=False
    ).head(n_users)

    denominator = movie_ratings['Similarity'].sum()

    if denominator == 0:
        return np.nan

    predicted_rating = (
        movie_ratings['Rating'] *
        movie_ratings['Similarity']
    ).sum() / denominator

    return predicted_rating
```

```python
predictions = []

for _, row in test_df.iterrows():

    predicted = predict_rating(
        row['User_ID'],
        row['Movie_ID']
    )

    if not np.isnan(predicted):

        predictions.append({
            'Actual': row['Rating'],
            'Predicted': predicted
        })

prediction_df = pd.DataFrame(predictions)

prediction_df.head()
```

```python
rmse = np.sqrt(
    mean_squared_error(
        prediction_df['Actual'],
        prediction_df['Predicted']
    )
)

print("RMSE:", rmse)
```

```python
def cosine_only_recommend(user_id, n=5):

    if user_id not in train_matrix.index:
        return pd.DataFrame(
            columns=['Movie_ID', 'Movie_Title', 'Score']
        )

    similarities = train_similarity_df.loc[user_id].drop(
        user_id,
        errors='ignore'
    )

    similar_users = similarities.sort_values(
        ascending=False
    ).head(10)

    recommendation_data = []

    for similar_user, similarity_score in similar_users.items():

        ratings = train_df[
            train_df['User_ID'] == similar_user
        ]

        for _, row in ratings.iterrows():

            recommendation_data.append({
                'Movie_ID': row['Movie_ID'],
                'Movie_Title': row['Movie_Title'],
                'Rating': row['Rating'],
                'Similarity': similarity_score
            })

    recommendation_df = pd.DataFrame(recommendation_data)

    if recommendation_df.empty:
        return recommendation_df

    recommendation_df['Score'] = (
        recommendation_df['Rating'] *
        recommendation_df['Similarity']
    )

    already_rated = set(
        train_df[
            train_df['User_ID'] == user_id
        ]['Movie_ID']
    )

    recommendation_df = recommendation_df[
        ~recommendation_df['Movie_ID'].isin(already_rated)
    ]

    return (
        recommendation_df
        .groupby(
            ['Movie_ID', 'Movie_Title'],
            as_index=False
        )['Score']
        .mean()
        .sort_values(
            'Score',
            ascending=False
        )
        .head(n)
    )
```

```python
cosine_recommendations = cosine_only_recommend(
    user_id=157,
    n=5
)

cosine_recommendations
```

```python
hybrid_recommendations = recommend_from_train(
    user_id=157,
    n=5
)

cosine_recommendations = cosine_only_recommend(
    user_id=157,
    n=5
)

print("K-Means + Cosine:")
display(hybrid_recommendations)

print("Cosine Only:")
display(cosine_recommendations)
```

```python
cosine_recommended_movies = cosine_recommendations['Movie_ID'].tolist()

relevant_movies = test_df[
    (test_df['User_ID'] == 157) &
    (test_df['Rating'] >= 7)
]['Movie_ID'].tolist()

cosine_precision = precision_at_k(
    cosine_recommended_movies,
    relevant_movies,
    k=5
)

cosine_recall = recall_at_k(
    cosine_recommended_movies,
    relevant_movies,
    k=5
)

print("Cosine Only Precision@5:", cosine_precision)
print("Cosine Only Recall@5:", cosine_recall)
```

```python
def predict_rating_cosine_only(user_id, movie_id, n_users=10):
    if user_id not in train_matrix.index:
        return np.nan

    if movie_id not in train_matrix.columns:
        return np.nan

    similarities = train_similarity_df.loc[user_id].drop(
        user_id,
        errors='ignore'
    )

    movie_ratings = train_df[
        (train_df['Movie_ID'] == movie_id) &
        (train_df['User_ID'].isin(similarities.index))
    ].copy()

    if movie_ratings.empty:
        return np.nan

    movie_ratings['Similarity'] = movie_ratings['User_ID'].map(
        similarities
    )

    movie_ratings = movie_ratings[
        movie_ratings['Similarity'] > 0
    ]

    if movie_ratings.empty:
        return np.nan

    movie_ratings = movie_ratings.sort_values(
        'Similarity',
        ascending=False
    ).head(n_users)

    denominator = movie_ratings['Similarity'].sum()

    if denominator == 0:
        return np.nan

    predicted_rating = (
        movie_ratings['Rating'] *
        movie_ratings['Similarity']
    ).sum() / denominator

    return predicted_rating
```

```python
cosine_predictions = []

for _, row in test_df.iterrows():

    predicted = predict_rating_cosine_only(
        row['User_ID'],
        row['Movie_ID']
    )

    if not np.isnan(predicted):
        cosine_predictions.append({
            'Actual': row['Rating'],
            'Predicted': predicted
        })

cosine_prediction_df = pd.DataFrame(cosine_predictions)

cosine_rmse = np.sqrt(
    mean_squared_error(
        cosine_prediction_df['Actual'],
        cosine_prediction_df['Predicted']
    )
)

print("Cosine Only RMSE:", cosine_rmse)
```

```python
comparison = pd.DataFrame({
    'Model': [
        'Cosine Only',
        'K-Means + Cosine'
    ],
    'Precision@5': [
        cosine_precision,
        precision
    ],
    'Recall@5': [
        cosine_recall,
        recall
    ],
    'RMSE': [
        cosine_rmse,
        rmse
    ]
})

comparison
```

```python
# Evaluate recommendation performance for multiple users

evaluation_results = []

test_users = test_df['User_ID'].unique()

for user_id in test_users:

    recommendations = recommend_from_train(
        user_id=user_id,
        n=5
    )

    if recommendations.empty:
        continue

    recommended_movies = recommendations['Movie_ID'].tolist()

    relevant_movies = test_df[
        (test_df['User_ID'] == user_id) &
        (test_df['Rating'] >= 7)
    ]['Movie_ID'].tolist()

    if len(relevant_movies) == 0:
        continue

    user_precision = precision_at_k(
        recommended_movies,
        relevant_movies,
        k=5
    )

    user_recall = recall_at_k(
        recommended_movies,
        relevant_movies,
        k=5
    )

    evaluation_results.append({
        'User_ID': user_id,
        'Precision@5': user_precision,
        'Recall@5': user_recall
    })

overall_evaluation = pd.DataFrame(evaluation_results)

print("Number of evaluated users:", len(overall_evaluation))

print(
    "Average Precision@5:",
    overall_evaluation['Precision@5'].mean()
)

print(
    "Average Recall@5:",
    overall_evaluation['Recall@5'].mean()
)

overall_evaluation.head()
```

```python
plt.figure(figsize=(7, 5))

metrics = ['Precision@5', 'Recall@5']
values = [
    overall_evaluation['Precision@5'].mean(),
    overall_evaluation['Recall@5'].mean()
]

plt.bar(metrics, values)

plt.title('Overall Recommendation Performance')
plt.ylabel('Score')
plt.ylim(0, max(values) * 1.3)

plt.show()
```
